```python
# 1. What is a Decision Tree?

# A Decision Tree is a supervised machine-learning model used for classification and regression.
# It splits the data into branches based on conditions to predict a target value.

# A tree consists of:

# Root Node – The starting feature for splitting

# Internal Nodes – Feature-based decision points

# Leaf Nodes – Final prediction (class label or value)

# Branches – Outcomes of a decision
```

## Project Objective

This project is a small practice-based machine learning project
focused on understanding how Decision Tree models work.

It uses a simple dataset to predict whether an employee’s salary
is greater than 100K based on company, job role, and education level.

The main goal is to learn the basic steps of building,
training, and evaluating a classification model using Python.

```python
import pandas as pd
```

```python
df = pd.read_csv('salaries.csv')
```

```python
inputs = df.drop('salary_more_then_100k',axis='columns')
```

```python
target = df['salary_more_then_100k']
```

```python
from sklearn.preprocessing import LabelEncoder # LabelEncoder converts categorical labels into numeric values,other option OneHotEncoder
lbl_company = LabelEncoder()
lbl_job = LabelEncoder()
lbl_degree = LabelEncoder()
```

```python
inputs['company_n'] = lbl_company.fit_transform(inputs['company'])
inputs['job_n'] = lbl_job.fit_transform(inputs['job'])
inputs['degree_n'] = lbl_degree.fit_transform(inputs['degree'])
```

```python
inputs
```

```python
inputs_numeric = inputs.drop(['company','job','degree'],axis='columns') # Keeping only numeric columns
```

```python
inputs_numeric
```

```python
target
```

```python
# Splitting data into training/testing sets:
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    inputs_numeric, target, test_size=0.2, random_state=42
)
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier(random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
# DecisionTreeClassifier(class_weight=None, criterion='gini', max_depth=None,
           # max_features=None, max_leaf_nodes=None,
           # min_impurity_decrease=0.0, min_impurity_split=None,
           # min_samples_leaf=1, min_samples_split=2,
           # min_weight_fraction_leaf=0.0, presort=False, random_state=None,
           # splitter='best')
```

```python
accuracy = model.score(X_test, y_test)
```

```python
print("Model Accuracy:", accuracy)
```

```python
# Making simple predictions using numeric values

# Format: [company_n, job_n, degree_n]

# Example 1: Google, Computer Programmer, Master's
prediction1 = model.predict([[2, 1, 1]])

# Example 2: ABC Pharma, Sales Executive, Bachelor's
prediction2 = model.predict([[0, 2, 0]])

# Example 3: facebook, computer programmer, masters's
prediction3 = model.predict([[1, 1 , 1]])

print(prediction1)
print(prediction2)
print(prediction3)

# 2 out of 3 correct → ~66% → matches your ~75% accuracy.
```

### Final Conclusion:
The model achieved around 75% accuracy on the test data.
Because the dataset is very small, the predictions are not always reliable.

Overall, this project helped in understanding the basic steps
of building, training, and using a machine learning classification model.
